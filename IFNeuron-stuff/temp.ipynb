```python
import h5py
```

```python
with h5py.File('test-1000-ms.hdf5', 'r') as f:
    # List all trial groups
    print("Trials in file:", list(f.keys()))

    # Access a specific trial group (e.g., 'trial_0')
    trial = f['trial_1']
    print(trial)

    # Get the column names
    columns = trial['data'].attrs['columns']
    print("Columns in dataset:", columns)

    # Access the dataset inside the trial group
    data = trial['data'][:]
    print("First 5 rows of data:")
    print(data[:5])

    # Extract all data for that column (all timesteps)
    neuron_a_spike_train = data[:, 1]
    print("Neuron_A spike train:", neuron_a_spike_train)
```

```python
# Find the index for the desired column
col_idx = list(columns).index("Neuron_A_spike_train")

# Extract all data for that column (all timesteps)
neuron_a_spike_train = data[:, col_idx]

print("Neuron_A spike train:", neuron_a_spike_train)
```
