```python
from NeuronNetwork import *
from utils import generate_random_stimulation_times_ms
from NeuronDataWriter import NeuronDataWriter
import numpy as np
```

```python
dataset_identifier = "b"
num_trials = 10                     # number of trials to run
simulation_run_time_ms = 180000     # simulated run time
num_stims = 100                     # number of neuron stimulations
min_time_between_stims_ms = 100     # min amount of time between stimulus

# I do this because I want to generate the same random stimulation times for each trial
neuron_a_input_times = []
neuron_b_input_times = []

for i in range(num_trials):
    neuron_a_input_times.append(generate_random_stimulation_times_ms(num_stims, simulation_run_time_ms, min_time_between_stims_ms))
    neuron_b_input_times.append(generate_random_stimulation_times_ms(num_stims, simulation_run_time_ms, min_time_between_stims_ms))
```

```python
writer = NeuronDataWriter(f"{dataset_identifier}Ground_Truth_data_{num_trials}trials_{num_stims}stims_{simulation_run_time_ms}ms.hdf5")

for trial_idx in range(num_trials):
    neuron_network = NeuronNetwork('Ground_Truth_System')
    input_stimulus_a = neuron_a_input_times[trial_idx].copy()
    input_stimulus_b = neuron_b_input_times[trial_idx].copy()

    neuron_network.add_neuron('Neuron_A')  # input neuron
    neuron_network.add_neuron('Neuron_B')  # input neuron
    neuron_network.add_neuron('Neuron_C')
    neuron_network.add_neuron('Neuron_D')
    neuron_network.add_neuron('Neuron_E')

    neuron_network.add_neuron_connection('Neuron_A', 'Neuron_C', .7)    # Neuron_A -> Neuron_C
    neuron_network.add_neuron_connection('Neuron_B', 'Neuron_C', .7)    # Neuron_B -> Neuron_C

    neuron_network.add_neuron_connection('Neuron_A', 'Neuron_D', 1.0)   # Neuron_A -> Neuron_D
    neuron_network.add_neuron_connection('Neuron_B', 'Neuron_D', 1.0)   # Neuron_B -> Neuron_D

    neuron_network.add_neuron_connection('Neuron_D', 'Neuron_E', 1.0)   # Neuron_D -> Neuron_E
    neuron_network.add_neuron_connection('Neuron_C', 'Neuron_E', -1.0)  # Neuron_C -> Neuron_E
    
    neuron_network.set_direct_stimulation_time_ms('Neuron_A', input_stimulus_a)

    neuron_network.set_direct_stimulation_time_ms('Neuron_B', input_stimulus_b)

    neuron_network.run_simulation(simulation_run_time_ms, True)
    writer.write_trial(neuron_network, trial_idx)

print("Done!")
```
